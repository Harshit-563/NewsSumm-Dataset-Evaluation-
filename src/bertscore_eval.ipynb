```python
import pandas as pd
import torch
from bert_score import score

# Check if CUDA is available for acceleration
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Using device: {device}')

# ======================
# LOAD DATA
# ======================
# Ensure the file 'NewsSumm++.xlsx' is uploaded to your Colab environment
try:
    df = pd.read_excel('/content/drive/MyDrive/NewsSumm++.xlsx')

    # 🔥 SMALL SAMPLE ONLY
    df = df.sample(min(200, len(df)), random_state=42)

    articles = df['article'].astype(str).tolist()
    summaries = df['summary'].astype(str).tolist()

    # ======================
    # BERTSCORE
    # ======================
    print('Calculating BERTScores...')
    P, R, F1 = score(
        summaries,
        articles,
        lang='en',
        batch_size=8,
        device=device
    )

    df['bertscore_f1'] = F1.cpu().numpy()

    print('Average BERTScore:', df['bertscore_f1'].mean())

    # Save results
    output_file = 'NewsSumm++_bertscore.xlsx'
    df.to_excel(output_file, index=False)
    print(f'Results saved to {output_file}')

except FileNotFoundError:
    print("Error: 'NewsSumm++.xlsx' not found. Please upload the file to the Files pane on the left.")
```

```python
import pandas as pd
import torch
from bert_score import score

# Check if CUDA is available for acceleration
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Using device: {device}')

# ======================
# LOAD DATA
# ======================
# Ensure the file 'NewsSumm.xlsx' is uploaded to your Colab environment
try:
    df = pd.read_excel('/content/drive/MyDrive/NewsSumm.xlsx')

    # 🔥 SMALL SAMPLE ONLY
    df.columns = df.columns.str.strip()

    print("Columns:", df.columns)

    # rename properly
    df = df.rename(columns={
        'article_text': 'article',
        'human_summary': 'summary'
    })
    df = df.sample(min(200, len(df)), random_state=42)

    articles = df['article'].astype(str).tolist()
    summaries = df['summary'].astype(str).tolist()

    # ======================
    # BERTSCORE
    # ======================
    print('Calculating BERTScores...')
    P, R, F1 = score(
        summaries,
        articles,
        lang='en',
        batch_size=8,
        device=device
    )

    df['bertscore_f1'] = F1.cpu().numpy()

    print('Average BERTScore:', df['bertscore_f1'].mean())

    # Save results
    output_file = 'NewsSumm_bertscore.xlsx'
    df.to_excel(output_file, index=False)
    print(f'Results saved to {output_file}')

except FileNotFoundError:
    print("Error: 'NewsSumm.xlsx' not found. Please upload the file to the Files pane on the left.")
```
